```python
connections = set()

def firewall(packet_type, connection_id):
    if packet_type == "outbound":
        connections.add(connection_id)
        print("Outbound packet allowed")

    elif packet_type == "inbound":
        if connection_id in connections:
            print("Return traffic allowed")
            connections.remove(connection_id)
        else:
            print("Unsolicited inbound packet blocked")


firewall("outbound", 101)
firewall("inbound", 101)
firewall("inbound", 202)

```
