```python
def classify_alert(alert):
    alert = alert.lower()

    if "ransomware" in alert or "data breach" in alert:
        return "P1 - Critical"
    elif "malware" in alert or "unauthorized access" in alert:
        return "P2 - High"
    elif "failed login" in alert or "suspicious activity" in alert:
        return "P3 - Medium"
    else:
        return "P4 - Low"


alert = input("Enter SOC alert description: ")
print("Alert Priority:", classify_alert(alert))
```
