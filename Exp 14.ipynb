```python
# Collect raw events from multiple sources
events = [
    {"source": "Firewall", "event": "failed_login"},
    {"source": "Server", "event": "failed_login"},
    {"source": "IDS", "event": "malware_detected"}
]

# Normalize events
normalized = []
for e in events:
    normalized.append({
        "source": e["source"],
        "event": e["event"].lower()
    })

print("Normalized Events:", normalized)

# Correlate events
failed = sum(1 for e in normalized
             if e["event"] == "failed_login")

malware = any(e["event"] == "malware_detected"
              for e in normalized)

# Raise alert
if failed >= 2 or malware:
    print("ALERT: Suspicious activity detected!")
else:
    print("No security alert")
```
