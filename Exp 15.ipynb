```python
def classify_incident(description):
    description = description.lower()

    if "phishing" in description or "fake email" in description:
        return "Phishing"
    elif "unpatched" in description or "outdated software" in description:
        return "Unpatched Vulnerability"
    elif "misconfiguration" in description or "wrong settings" in description:
        return "Misconfiguration"
    elif "employee" in description or "insider" in description:
        return "Insider Threat"
    elif "third-party" in description or "vendor" in description:
        return "Supply-Chain Compromise"
    else:
        return "Unknown Root Cause"


incident = input("Enter incident description: ")
print("Root Cause:", classify_incident(incident))
```
