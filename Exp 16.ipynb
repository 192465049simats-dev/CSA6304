```python
layers = ["Firewall", "IDS", "Antivirus", "Authentication"]

attack_blocked = False

for layer in layers:
    result = input(layer + " blocked the attack? (yes/no): ").lower()

    if result == "yes":
        print(layer, "blocked the attack!")
        attack_blocked = True
        break
    else:
        print("Attack passed through", layer)

if attack_blocked:
    print("System Protected")
else:
    print("Attack reached the target!")
```
