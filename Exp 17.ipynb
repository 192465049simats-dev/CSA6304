```python
description = input("Enter security control description: ").lower()

if any(word in description for word in ["policy", "risk", "supplier", "compliance"]):
    domain = "Organizational Controls"
elif any(word in description for word in ["employee", "training", "awareness", "staff"]):
    domain = "People Controls"
elif any(word in description for word in ["building", "door", "cctv", "physical"]):
    domain = "Physical Controls"
elif any(word in description for word in ["firewall", "encryption", "network", "software"]):
    domain = "Technological Controls"
else:
    domain = "Unknown Control Domain"

print("ISO 27001 Annex A Domain:", domain)
```
