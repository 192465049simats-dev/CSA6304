```python
events = [
    {"user": "John", "status": "failed"},
    {"user": "Alex", "status": "success"},
    {"user": "John", "status": "failed"},
    {"user": "Alex", "status": "success"},
    {"user": "John", "status": "success"}
]

# Search filter
print("Failed Login Events:")
for event in events:
    if event["status"] == "failed":
        print(event)

# Stats count by field
print("\nLogin Count by Status:")
count = {}

for event in events:
    status = event["status"]
    count[status] = count.get(status, 0) + 1

for status, total in count.items():
    print(status, ":", total)
```
