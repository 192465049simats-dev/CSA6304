```python

def identify_stage(activity):
    activity = activity.lower()

    stages = {
        "Direction": ["requirements", "objectives", "planning"],
        "Collection": ["collect", "gather", "data", "logs"],
        "Processing": ["clean", "filter", "organize", "format"],
        "Analysis": ["analyze", "investigate", "identify", "patterns"],
        "Dissemination": ["share", "report", "distribute", "communicate"],
        "Feedback": ["feedback", "review", "improve", "suggestions"]
    }

    for stage, keywords in stages.items():
        if any(word in activity for word in keywords):
            return stage

    return "Stage not identified"


activity = input("Enter activity description: ")
print("Lifecycle Stage:", identify_stage(activity))
```
